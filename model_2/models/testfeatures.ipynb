```python
import pandas as pd
import json
```

```python
# --- What features the XGBoost model actually trains on ---
with open("feature_list.json") as f:
    feature_config = json.load(f)

xgb_features = feature_config["xgboost_features"]
lstm_features = feature_config["lstm_features"]

print(f"XGBoost feature count: {len(xgb_features)}")
for f in xgb_features:
    print(" -", f)

print(f"\nLSTM feature count: {len(lstm_features)}")
for f in lstm_features:
    print(" -", f)
```
