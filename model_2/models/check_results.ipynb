```python
import pandas as pd

df = pd.read_csv("walk_forward_results.csv")

# Overview: how many folds per model/cadence
print("Fold counts:")
print(df.groupby(["model", "cadence"]).size())
print()

# The actual comparison: mean performance across folds
summary = df.groupby(["model", "cadence"])[["accuracy", "f1_macro", "log_loss"]].mean()
print("Mean performance across folds:")
print(summary.round(4))
print()

# Best cadence per model (by f1_macro, since accuracy alone is misleading
# with an imbalanced flat/up/down split)
for model in df["model"].unique():
    sub = summary.loc[model]
    best = sub["f1_macro"].idxmax()
    print(f"{model}: best cadence by f1_macro = {best} "
          f"(f1_macro={sub.loc[best, 'f1_macro']:.4f}, "
          f"accuracy={sub.loc[best, 'accuracy']:.4f}, "
          f"log_loss={sub.loc[best, 'log_loss']:.4f})")
```
